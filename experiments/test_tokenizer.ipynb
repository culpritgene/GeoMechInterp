```python
from geomechinterp.causal.mygpt import ALL_SYMBOLS, EXTRA_SYMBOLS
from geomechinterp.causal.mygpt import create_tokenize_function, SymbolTokenizer
from datasets import Dataset
import json


def load_dataset(data_or_path):
    """Load and format the dataset."""
    if isinstance(data_or_path, str):
        data = json.load(open(data_or_path))
    elif isinstance(data_or_path, dict):
        data = data_or_path
    else:
        raise ValueError(f"Invalid data type: {type(data_or_path)}")
    data = [
        {
            "text": k,  # pattern
            "generator": v["generator"],
            "stochastic": v["stochastic"],
            "sequence_of_features": v["sequence_of_features"],
        }
        for k, v in data.items()
    ]
    return Dataset.from_list(data)


data_path = "/Users/solar/Code/DL/Mechanistic_Interpretability/GeoMechInterp/data/all_patterns_and_generators_test.json"
# Create vocabulary and tokenizer
custom_vocab = ALL_SYMBOLS + EXTRA_SYMBOLS
tokenizer = SymbolTokenizer(custom_vocab)

# Load and process dataset
dataset = load_dataset(data_path)
tokenize_func = create_tokenize_function(tokenizer)
tokenized_dataset = dataset.map(tokenize_func, batched=True)
```

```python
from transformers import DataCollatorForLanguageModeling
from transformers.data.data_collator import DataCollatorMixin


class DataCollator(DataCollatorMixin):
    def __init__(self, dataset):
        self.dataset = dataset
        super().__init__()
        
    def __call__(self, idx: list[int] | int):
        return self.dataset[idx]['input_ids']

data_collator = DataCollator(tokenized_dataset)

data_collator([1,2,3])
```
