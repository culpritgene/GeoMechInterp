```python
from transformers import GPT2Config, GPT2LMHeadModel

def get_model_config(vocab_size):
    """Create GPT model configuration."""
    return GPT2Config(
        vocab_size=vocab_size,
        n_positions=128,
        n_ctx=128,
        n_embd=128,
        n_layer=6,
        n_head=4,
        use_cache=True,
        rotary_dimss=64,
        return_dict=True,
    )

config = get_model_config(vocab_size=50256)

model = GPT2LMHeadModel(config)

model
```
